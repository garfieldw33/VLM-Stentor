## json 2 vtt

```python
import json

# Load the result dictionary from the JSON file
with open('/Users/kesiyun/Desktop/00workspace/00sequenceresult.json', 'r') as f:
    loaded_result = json.load(f)

for key, value in loaded_result.items():
    print(key,': ',value)
```

```python
#R - resting, B - bending, A - ciliary alteration, C - contraction, D - detachment P

goldstd={
    'R':'R - resting',
    'B':'B - bending',
    'A':'A - ciliary alteration',
    'C':'C - contraction',
    'D':'D - detachment',
    'P':'P - pulse'
}


def convert_to_vtt(input_string):
    import re

    # Function to convert time format
    def convert_time(time_str):
        minutes, seconds = map(int, time_str.split(':'))
        return f"{minutes // 60:02}:{minutes % 60:02}:{seconds:02}.000"

    # Function to add 5 seconds to the time
    def add_five_seconds(time_str):
        minutes, seconds = map(int, time_str.split(':'))
        total_seconds = minutes * 60 + seconds + 5
        return f"{total_seconds // 3600:02}:{(total_seconds % 3600) // 60:02}:{total_seconds % 60:02}.000"

    # Initialize VTT content
    vtt_content = "WEBVTT\n\n"

    # Find all matches in the input string
    matches = re.findall(r'([A-Z])\((\d+:\d+)\)', input_string)

    # Add the initial "R" caption
    #vtt_content += "00:00:00.000 --> " + convert_time(matches[0][1]) + "\nR\n\n"
    vtt_content += "00:00:00.000 --> " + convert_time(matches[0][1]) + "\n"+goldstd['R']+"\n\n"
    
    # Iterate over matches and create VTT entries
    for i in range(len(matches) - 1):
        start_time = convert_time(matches[i][1])
        end_time = convert_time(matches[i + 1][1])
        #caption = matches[i][0]
        caption=goldstd[matches[i][0]]
        vtt_content += f"{start_time} --> {end_time}\n{caption}\n\n"

    ## Add the last caption with an estimated end time
    #last_start_time = convert_time(matches[-1][1])
    #vtt_content += f"{last_start_time} --> 01:00:00.000\n{matches[-1][0]}\n"
    
    # Add the last caption with an estimated end time (5 seconds later)
    last_start_time = convert_time(matches[-1][1])
    last_end_time = add_five_seconds(matches[-1][1])
    #vtt_content += f"{last_start_time} --> {last_end_time}\n{matches[-1][0]}\n" #
    vtt_content += f"{last_start_time} --> {last_end_time}\n{goldstd[matches[-1][0]]}\n" 

    return vtt_content



# Sample input strings
#input_string_1 = "R P(1:40) A(1:44)B(1:51)C(2:06)C(10:50) P(12:54) A(13:21)C(13:28)A(17:25)C(17:30)"
#input_string_2 = "R P(1:39) A(1:46)"
#input_string_3 = "R P(0:21) A(0:40)B(1:40) P(2:52) C(3:25)D(4:05)"

# Convert to VTT format and print the results
#print(convert_to_vtt(input_string_1))
#print(convert_to_vtt(input_string_2))
#print(convert_to_vtt(input_string_3))
```

```python


base_dir='/Users/kesiyun/Desktop/00workspace/videos/'
# Load the result dictionary from the JSON file
#with open(base_dir+'sequenceresult.json', 'r') as f:
    #loaded_result = json.load(f)

for key, value in loaded_result.items():
    #print(key,': ',value)
    
    filename=key+'.VTT'
    filedir=base_dir+filename
    VTTcontent=convert_to_vtt(value['timesec'])
    
    #print(filedir)
    with open(filedir, 'w') as f:
        f.write(VTTcontent)

    
#print()
#print()
```
