```python
import os
import json

import pandas as pd
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import ast
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix
from sklearn.metrics import accuracy_score, precision_score, f1_score
import numpy as np
from torch.utils.data import TensorDataset, DataLoader, random_split
from sklearn.metrics import precision_score, recall_score, f1_score, accuracy_score


# Utility functions
def read_json(path):
    with open(path, 'r') as f:
        return json.load(f)

def save_json(d, path):
    with open(path, 'w') as f:
        json.dump(d, f, indent=4)
        
def print_loader_metadata(train_loader, test_loader):
    def get_loader_stats(loader, name):
        label_counts = {}
        feature_shapes = {}
        total_samples = 0

        for batch in loader:
            features, labels = batch
            total_samples += len(labels)

            for feature, label in zip(features, labels):
                label_val = label.item()
                label_counts[label_val] = label_counts.get(label_val, 0) + 1

                shape = tuple(feature.shape)
                feature_shapes[shape] = feature_shapes.get(shape, 0) + 1

        print(f"\n{name} Loader:")
        print(f"  Total samples: {total_samples}")
        print(f"  Total batches (batch_size={loader.batch_size}): {len(loader)}")
        print(f"  Feature shape(s):")
        for shape, count in feature_shapes.items():
            print(f"    Shape {shape}: {count} samples")
        print(f"  Label distribution:")
        for label, count in sorted(label_counts.items()):
            print(f"    Label {label}: {count} samples")

    get_loader_stats(train_loader, "Training")
    get_loader_stats(test_loader, "Testing")


device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

def visualize_c(video_name):
    visualized_str=''
    feature_list_load_1 = torch.load(os.path.join(features_dir, video_name + '.pt'))
    for i in range (len(feature_list_load_1)):
        if (feature_list_load_1[i][1]==1):
            visualized_str=visualized_str+'I'
        else:
            visualized_str=visualized_str+'_'
    return visualized_str

def visualize_c_tensor(video_name):
    visualized_str=''

    feature_sample = []
    label_sample = []
    feature_list_load_1 = torch.load(os.path.join(features_dir, video_name + '.pt'))
    for i in range(len(feature_list_load_1)):
        feature_this = feature_list_load_1[i][0]  # shape: [1, 2304]
        label_this = feature_list_load_1[i][1]    # assuming label is stored here as an integer
        feature_sample.append(feature_this)
        label_sample.append(label_this)
    # Convert to tensors
    X_sample = torch.cat(feature_sample, dim=0)  # shape: [num_samples, 2304]
    y_sample = torch.tensor(label_sample).float().unsqueeze(1)  # shape: [num_samples, 1]
    
    for i in range (len(feature_list_load_1)):
        if (y_sample[i]==1):
            visualized_str=visualized_str+'I'
        else:
            visualized_str=visualized_str+'_'
    return visualized_str

def visualize_c_tensor_pred(video_name):
    visualized_str=''
    feature_sample = []
    label_sample = []
    feature_list_load_1 = torch.load(os.path.join(features_dir, video_name + '.pt'))
    for i in range(len(feature_list_load_1)):
        feature_this = feature_list_load_1[i][0]  # shape: [1, 2304]
        label_this = feature_list_load_1[i][1]    # assuming label is stored here as an integer
        feature_sample.append(feature_this)
        label_sample.append(label_this)
    # Convert to tensors
    X_sample = torch.cat(feature_sample, dim=0)  # shape: [num_samples, 2304]
    y_sample = torch.tensor(label_sample).float().unsqueeze(1)  # shape: [num_samples, 1]
    
    model.eval()
    features = X_sample.to(device)
    outputs = model(features)
    preds = (outputs > 0.5).float().cpu()
    
    # Convert to flat arrays
    preds = torch.tensor(preds).view(-1).numpy()
    for i in range (len(feature_list_load_1)):
        if (preds[i]==1):
            visualized_str=visualized_str+'I'
        else:
            visualized_str=visualized_str+'_'
    return visualized_str


def get_prediction_string(video_name):
    feature_sample = []
    feature_list_load_1 = torch.load(os.path.join(features_dir, video_name + '.pt'))
    for i in range(len(feature_list_load_1)):
        feature_this = feature_list_load_1[i][0]
        feature_sample.append(feature_this)
    X_sample = torch.cat(feature_sample, dim=0)
    model.eval()
    features = X_sample.to(device)
    outputs = model(features)
    preds = (outputs > 0.5).float().cpu().view(-1).numpy()
    return ''.join(['I' if p == 1 else '_' for p in preds])

def get_ground_truth_string(video_name):
    feature_list_load_1 = torch.load(os.path.join(features_dir, video_name + '.pt'))
    return ''.join(['I' if item[1] == 1 else '_' for item in feature_list_load_1])
    

def get_prediction_probabilities(video_name):
    prob_pairs = []
    feature_sample = []
    label_sample = []

    feature_list_load_1 = torch.load(os.path.join(features_dir, video_name + '.pt'))
    for i in range(len(feature_list_load_1)):
        feature_this = feature_list_load_1[i][0]  # shape: [1, 2304]
        label_this = feature_list_load_1[i][1]
        feature_sample.append(feature_this)
        label_sample.append(label_this)

    # Convert to tensors
    X_sample = torch.cat(feature_sample, dim=0)  # shape: [num_samples, 2304]
    y_sample = torch.tensor(label_sample).float().unsqueeze(1)

    model.eval()
    features = X_sample.to(device)

    with torch.no_grad():
        if NEED_sigmoid_outout:
            outputs = torch.sigmoid(model(features))  # raw logits → sigmoid
        else:
            outputs = model(features)  # already sigmoid output

        probs = outputs.cpu().numpy().flatten()
        for p in probs:
            prob_pairs.append((1 - p, p))  # (prob of class 0, prob of class 1)

    return prob_pairs

```

```python
#get prob pairs
#fuse()
def prob_pair_to_vtt(prob_pairs):
    vtt_lines = ["WEBVTT\n"]
    for i, (prob_0, prob_1) in enumerate(prob_pairs):
        start_time = f"00:00:{i:02d}.000"
        end_time = f"00:00:{i+1:02d}.000"
        label = "C" if prob_1 > prob_0 else "X"
        line = f"{start_time} --> {end_time}\n{prob_0:.4f}, {label}, {prob_1:.4f}\n"
        vtt_lines.append(line)
    return "\n".join(vtt_lines)


#base_dir='/Users/kesiyun/Desktop/00workspace/videos/'
vtt_output_dir='/Users/kesiyun/Desktop/00workspace/VTTs/0/'
test_set_videos=['4E', '12B', '17I', '4C', '16A', '4A']
#videoname='17I' #333 '4E' '12B'   #22 '17I' '4C' # 1 16A
for videoname in test_set_videos:
    prob_pair_pred = get_prediction_probabilities(videoname)
    VTTcontent = prob_pair_to_vtt(prob_pair_pred)
    
    filename=videoname+'.out.VTT'
    filedir=vtt_output_dir+filename    
    with open(filedir, 'w') as f:
        f.write(VTTcontent)



    
#print()
#print()
```

```python
#IoU
# ----- IoU and Temporal Metrics -----
def extract_segments(sequence_str):
    segments = []
    start = None
    for i, char in enumerate(sequence_str):
        if char == 'I' and start is None:
            start = i
        elif char == '_' and start is not None:
            segments.append((start, i - 1))
            start = None
    if start is not None:
        segments.append((start, len(sequence_str) - 1))
    return segments

def compute_iou(pred_seg, gt_seg):
    pred_start, pred_end = pred_seg
    gt_start, gt_end = gt_seg
    intersection = max(0, min(pred_end, gt_end) - max(pred_start, gt_start) + 1)
    union = max(pred_end, gt_end) - min(pred_start, gt_start) + 1
    return intersection / union if union > 0 else 0

def compute_miou(pred_str, gt_str):
    pred_segments = extract_segments(pred_str)
    gt_segments = extract_segments(gt_str)
    if not pred_segments or not gt_segments:
        return 0.0
    iou_scores = []
    for pred_seg in pred_segments:
        best_iou = 0.0
        for gt_seg in gt_segments:
            iou = compute_iou(pred_seg, gt_seg)
            best_iou = max(best_iou, iou)
        iou_scores.append(best_iou)
    return sum(iou_scores) / len(iou_scores) if iou_scores else 0.0

def compute_temporal_precision_recall(pred_str, gt_str, iou_threshold=0.5):
    pred_segments = extract_segments(pred_str)
    gt_segments = extract_segments(gt_str)
    matched_gt = set()
    TP = 0
    for pred_seg in pred_segments:
        best_match = None
        best_iou = 0.0
        for idx, gt_seg in enumerate(gt_segments):
            iou = compute_iou(pred_seg, gt_seg)
            if iou > best_iou:
                best_iou = iou
                best_match = idx
        if best_iou >= iou_threshold:
            TP += 1
            matched_gt.add(best_match)
    FP = len(pred_segments) - TP
    FN = len(gt_segments) - len(matched_gt)
    precision = TP / (TP + FP) if (TP + FP) > 0 else 0.0
    recall = TP / (TP + FN) if (TP + FN) > 0 else 0.0
    f1 = 2 * precision * recall / (precision + recall) if (precision + recall) > 0 else 0.0
    return TP, FP, FN, precision, recall, f1

# ----- Model Definition -----
class LinearClassifier(nn.Module):
    def __init__(self, input_dim, use_sigmoid=True):
        super(LinearClassifier, self).__init__()
        self.linear = nn.Linear(input_dim, 1)
        self.use_sigmoid = use_sigmoid
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        x = self.linear(x)
        if self.use_sigmoid:
            x = self.sigmoid(x)
        return x

        
base_dir="/Users/kesiyun/Desktop/00workspace/savedmodels"

model_meta_dict=read_json(os.path.join(base_dir,'00model_meta.json'))
print('Select model to load:')
for key in model_meta_dict.keys():
    print(key,':',model_meta_dict[key]['description'])

test_set_videos=['4E', '12B', '17I', '4C', '16A', '4A']
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score, accuracy_score
import numpy as np

test_set_videos = ['4E', '12B', '17I', '4C', '16A', '4A']

all_preds = []
all_labels = []
video_metrics = {}

TP_c = FP_c = FN_c = TN_c = 0  # Global counts

for videoname in test_set_videos:
    # Get prediction and ground truth strings
    pred_str = get_prediction_string(videoname)
    gt_str = get_ground_truth_string(videoname)
    
    # Convert to binary arrays
    preds = np.array([1 if ch == 'I' else 0 for ch in pred_str])
    labels = np.array([1 if ch == 'I' else 0 for ch in gt_str])
    
    # Per-video metrics
    precision = precision_score(labels, preds, pos_label=1)
    recall = recall_score(labels, preds, pos_label=1)
    f1 = f1_score(labels, preds, pos_label=1)
    accuracy = accuracy_score(labels, preds)
    
    video_metrics[videoname] = {
        'precision': precision,
        'recall': recall,
        'f1': f1,
        'accuracy': accuracy
    }
    
    # Update global TP/FP/FN/TN
    for gt, pr in zip(labels, preds):
        if gt == 1 and pr == 1:
            TP_c += 1
        elif gt == 0 and pr == 1:
            FP_c += 1
        elif gt == 1 and pr == 0:
            FN_c += 1
        else:
            TN_c += 1
    
    # Accumulate for global metrics
    all_preds.extend(preds)
    all_labels.extend(labels)

# Global metrics
precision_C = TP_c / (TP_c + FP_c) if (TP_c + FP_c) > 0 else 0.0
recall_C = TP_c / (TP_c + FN_c) if (TP_c + FN_c) > 0 else 0.0
f1_C = (2 * precision_C * recall_C) / (precision_C + recall_C) if (precision_C + recall_C) > 0 else 0.0
accuracy_global = (TP_c + TN_c) / (TP_c + FP_c + FN_c + TN_c)

# Macro metrics
precision_nonC = TN_c / (TN_c + FN_c) if (TN_c + FN_c) > 0 else 0.0
recall_nonC = TN_c / (TN_c + FP_c) if (TN_c + FP_c) > 0 else 0.0
f1_nonC = (2 * precision_nonC * recall_nonC) / (precision_nonC + recall_nonC) if (precision_nonC + recall_nonC) > 0 else 0.0
macro_precision = (precision_C + precision_nonC) / 2
macro_f1 = (f1_C + f1_nonC) / 2

# ----- Output -----
print("\n--- Per-Video Metrics ---")
for vid, metrics in video_metrics.items():
    print(f"{vid}: Precision={metrics['precision']:.4f}, Recall={metrics['recall']:.4f}, "
          f"F1={metrics['f1']:.4f}, Acc={metrics['accuracy']:.4f}")

print("\n--- Detailed Evaluation on Test Set ---")
print(f"True Positives (C): {TP_c}")
print(f"False Positives (C): {FP_c}")
print(f"False Negatives (C): {FN_c}")
print(f"Precision (C): {precision_C:.4f}")
print(f"Recall (C): {recall_C:.4f}")
print(f"F1 Score (C): {f1_C:.4f}")
print(f"Accuracy: {accuracy_global:.4f}")
print(f"Macro Precision: {macro_precision:.4f}")
print(f"Macro F1: {macro_f1:.4f}")
```

```python
def update_prediction_with_state(pred_str, state_str, alpha=0.1, beta=1.2):
    """
    Update prediction string based on state string.
    
    Args:
        pred_str (str): Original prediction string of '_' and 'I'.
        state_str (str): State string of '_' and 'I'.
        alpha (float): Suppression factor when state = '_'.
        beta (float): Boost factor when state = 'I'.
    
    Returns:
        updated_str (str): Updated prediction string.
    """
    updated_chars = []
    min_len = min(len(pred_str), len(state_str))  # Only process overlapping part
    
    for i in range(min_len):
        pred_char = pred_str[i]
        state_char = state_str[i]
        
        if state_char == '_':
            # Suppress prediction: force '_'
            updated_chars.append('_')
        else:
            # Keep original prediction or boost
            updated_chars.append(pred_char)
    
    # If pred_str is longer, append remaining original predictions
    if len(pred_str) > min_len:
        updated_chars.extend(pred_str[min_len:])
    
    return ''.join(updated_chars)

pred_str = "___________IIIII_________________________________________IIIII____________I__________________________IIII_____________________________________________________________________________________________________________________________________________________________________________________________________IIIII___II__IIIII______________________________________________________________________________________________________________________________________________________________________________________________________________________________________________________________________________________"
state_str = "_____IIIII______________________________IIIIIIIIIIIIIIIIIIII_______________IIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIII_____IIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIII_____IIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIIII_____IIIIIIIIIIIIIIIIIIIIIIIIIIIIII"
#print(pred_str)  
#print(update_prediction_with_state(pred_str, state_str))  
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score, accuracy_score
import numpy as np

test_set_videos = ['4E', '12B', '17I', '4C', '16A', '4A']
state_strings=read_json("/Users/kesiyun/Desktop/00workspace/bstatemodels/state_strings.json")

all_preds = []
all_labels = []
video_metrics = {}

TP_c = FP_c = FN_c = TN_c = 0  # Global counts

for videoname in test_set_videos:
    
    # Get prediction and ground truth strings
    pred_str = get_prediction_string(videoname)

    state_str=state_strings[videoname]
    pred_str = update_prediction_with_state(pred_str, state_str,0.1,1.5)
    
    gt_str = get_ground_truth_string(videoname)
    
    # Convert to binary arrays
    preds = np.array([1 if ch == 'I' else 0 for ch in pred_str])
    labels = np.array([1 if ch == 'I' else 0 for ch in gt_str])
    
    # Per-video metrics
    precision = precision_score(labels, preds, pos_label=1)
    recall = recall_score(labels, preds, pos_label=1)
    f1 = f1_score(labels, preds, pos_label=1)
    accuracy = accuracy_score(labels, preds)
    
    video_metrics[videoname] = {
        'precision': precision,
        'recall': recall,
        'f1': f1,
        'accuracy': accuracy
    }
    
    # Update global TP/FP/FN/TN
    for gt, pr in zip(labels, preds):
        if gt == 1 and pr == 1:
            TP_c += 1
        elif gt == 0 and pr == 1:
            FP_c += 1
        elif gt == 1 and pr == 0:
            FN_c += 1
        else:
            TN_c += 1
    
    # Accumulate for global metrics
    all_preds.extend(preds)
    all_labels.extend(labels)

# Global metrics
precision_C = TP_c / (TP_c + FP_c) if (TP_c + FP_c) > 0 else 0.0
recall_C = TP_c / (TP_c + FN_c) if (TP_c + FN_c) > 0 else 0.0
f1_C = (2 * precision_C * recall_C) / (precision_C + recall_C) if (precision_C + recall_C) > 0 else 0.0
accuracy_global = (TP_c + TN_c) / (TP_c + FP_c + FN_c + TN_c)

# Macro metrics
precision_nonC = TN_c / (TN_c + FN_c) if (TN_c + FN_c) > 0 else 0.0
recall_nonC = TN_c / (TN_c + FP_c) if (TN_c + FP_c) > 0 else 0.0
f1_nonC = (2 * precision_nonC * recall_nonC) / (precision_nonC + recall_nonC) if (precision_nonC + recall_nonC) > 0 else 0.0
macro_precision = (precision_C + precision_nonC) / 2
macro_f1 = (f1_C + f1_nonC) / 2

# ----- Output -----
print("\n--- Per-Video Metrics ---")
for vid, metrics in video_metrics.items():
    print(f"{vid}: Precision={metrics['precision']:.4f}, Recall={metrics['recall']:.4f}, "
          f"F1={metrics['f1']:.4f}, Acc={metrics['accuracy']:.4f}")

print("\n--- Detailed Evaluation on Test Set ---")
print(f"True Positives (C): {TP_c}")
print(f"False Positives (C): {FP_c}")
print(f"False Negatives (C): {FN_c}")
print(f"Precision (C): {precision_C:.4f}")
print(f"Recall (C): {recall_C:.4f}")
print(f"F1 Score (C): {f1_C:.4f}")
print(f"Accuracy: {accuracy_global:.4f}")
print(f"Macro Precision: {macro_precision:.4f}")
print(f"Macro F1: {macro_f1:.4f}")
```

```python
import string

def visualize_segments_with_letters(pred_str):
    # Extract segments
    segments = extract_segments(pred_str)
    
    # Prepare letters (A-Z, then AA, AB if needed) 
    letters = list(string.ascii_uppercase)
    letters += letters
    letters += letters
    letters += letters #let s just repeat
    #print(letters)
    while len(letters) < len(segments):
        # Extend with double letters if more than 26 segments
        letters += [a+b for a in string.ascii_uppercase for b in string.ascii_uppercase]
    
    # Initialize visualization as list of underscores
    vis_list = ['_'] * len(pred_str)
    
    # Fill in letters for each segment
    for idx, (start, end) in enumerate(segments):
        for i in range(start, end + 1):
            vis_list[i] = letters[idx]
    
    return ''.join(vis_list)

# Example usage:
pred_str = "_____IIIII_____IIII____I__III____I"
visualized_letters = visualize_segments_with_letters(pred_str)
print(visualized_letters)
```
