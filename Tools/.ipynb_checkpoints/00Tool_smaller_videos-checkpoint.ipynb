```python
import subprocess
import os

def batch_convert_avi_to_mp4(input_folder, output_folder, crf=28, preset="fast"):
    """
    Batch convert all AVI files in input_folder to MP4 in output_folder,
    resizing to half the original dimensions.
    
    Parameters:
    - input_folder: str, path to folder containing AVI files
    - output_folder: str, path to folder where MP4 files will be saved
    - crf: int, quality factor (lower = better quality, bigger file; 28 is good for small size)
    - preset: str, speed/efficiency tradeoff ("ultrafast", "fast", "medium", "slow")
    """
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)
    
    avi_files = [f for f in os.listdir(input_folder) if f.lower().endswith(".avi")]
    
    if not avi_files:
        print("No AVI files found in the input folder.")
        return
    
    for avi_file in avi_files:
        input_path = os.path.join(input_folder, avi_file)
        output_path = os.path.join(output_folder, os.path.splitext(avi_file)[0] + ".mp4")
        
        # FFmpeg command with scale filter (half width and height)
        command = [
            "ffmpeg",
            "-i", input_path,
            "-vf", "scale=iw/4:ih/4",  # Resize to half original dimensions /2, /4
            "-vcodec", "libx264",
            "-crf", str(crf),
            "-preset", preset,
            "-pix_fmt", "yuv420p",
            "-movflags", "+faststart",
            output_path
        ]
        
        print(f"Converting: {avi_file} → {os.path.basename(output_path)} (resized)")
        subprocess.run(command, check=True)
    
    print("✅ Batch conversion with resizing complete!")

```

```python
# Example usage:
batch_convert_avi_to_mp4(
    "/Volumes/Green SSD/00Workspace_portable/videos",
    "/Volumes/Green SSD/00Workspace_portable/videos250",
    crf=30,  # Higher CRF = smaller file
    preset="fast"
)
```
